### Download, and first visualizations:

```python
def load_housing_data():
    tarball_path = Path("datasets/housing.tgz")
    if not tarball_path.is_file():
        Path("datasets").mkdir(parents=True, exist_ok=True)
        url = "https://github.com/ageron/data/raw/main/housing.tgz"
        urllib.request.urlretrieve(url, tarball_path)
        with tarfile.open(tarball_path) as housing_tarball:
            housing_tarball.extractall(path="datasets")
    return pd.read_csv(Path("datasets/housing/housing.csv"))
```

### Generating train and test set with random values:

```python
# It works, but in other program runnings, the model will know all the Dataset. We dont want it
def suffle_and_split_data(data, test_ratio):
    """
    data: The worst pd.DataFrame
    test_ratio: The % separated for the test set
    """
    suffled_indices = np.random.permutation(len(data)) # Returns random indices (0-len(data) - 1)
    test_set_size = int(len(data) * test_ratio)
    test_indices = suffled_indices[:test_set_size] # Gets only 20% of the random indices for the test set
    train_indices = suffled_indices[test_set_size:] # Gets 80% of the random indices for the training set
    return data.iloc[train_indices], data.iloc[test_indices]
```

```python
train_set, test_set = suffle_and_split_data(housing, 0.2)
```

### Generating train and test set with random values:

```python
def is_id_in_test_set(identifier, test_ratio):
    """
    Returns if the generated hash of [identifier] is smaller than [test_ratio]% of its maximum potential lenght.
    It does not guarantee if exactly 20% os the worst data will be separated for the test set,
    but there is not much difference in practice
    """
    return crc32(np.int64(identifier)) < test_ratio * 2**32
```

```python
def split_data_with_id_hash(data, test_ratio, id_column):
    ids = data[id_column]
    in_test_set = ids.apply(lambda _id: is_id_in_test_set(_id, test_ratio))
    return data.loc[~in_test_set], data.loc[in_test_set]
```

```python
housing_with_id["id"] = housing["longitude"] * 1000 + housing["latitude"] # Calculate stable identifiers
```

```python
train_set, test_set = split_data_with_id_hash(housing_with_id, 0.2, "id")
```

### Cutting median income into categories (stratified sampling) and visualizing it:

```python
# New categorical column: 'income_cat'
# Labels:
# 1 --> 0. - 1.5
# 2 --> 1.5 - 3.0
# 3 --> 3.0 - 4.5
# 4 --> 4.5 - 6.0
# 5 --> 6.0 - ...
housing["income_cat"] = pd.cut(housing["median_income"], bins=[0., 1.5, 3.0, 4.5, 6., np.inf], labels=[1, 2, 3, 4, 5])
```

```python
housing["income_cat"].value_counts().sort_index().plot.bar(rot=0, grid=True)
plt.xlabel("Income category")
plt.ylabel("Number os distrincts")
plt.show()
```

#### With SciKit-Learn:

```python
splitter = StratifiedShuffleSplit(n_splits=10, test_size=0.2, random_state=42)
```

```python
strat_splits = []

for train_index, test_index in splitter.split(housing, housing["income_cat"]):
    strat_train_set_n = housing.iloc[train_index]
    strat_test_set_n = housing.iloc[test_index]
    strat_splits.append([strat_train_set_n, strat_test_set_n])
```

```python
strat_train_set, strat_test_set = strat_splits[0] # We can just get the first division
```

```python
# Or we can pass the 'stratify' into sklearn.model_selection.train_test_split
strat_train_set, strat_test_set = train_test_split(housing, test_size=0.2, stratify=housing["income_cat"], random_state=42)
```

```python
strat_test_set["income_cat"].value_counts() / len(strat_test_set)
```

### Deleting new column after classification studies:

```python
for set_ in (strat_train_set, strat_test_set):
    set_.drop("income_cat", axis=1, inplace=True)
```

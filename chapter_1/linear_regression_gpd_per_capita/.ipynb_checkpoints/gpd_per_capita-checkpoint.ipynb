```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
```

```python
# Preparing the data
data_root = "https://github.com/ageron/data/raw/main/"
lifesat = pd.read_csv(data_root + "lifesat/lifesat.csv")
```

```python
x = lifesat[["GDP per capita (USD)"]].values
y = lifesat[["Life satisfaction"]].values
```

```python
lifesat.plot(kind="scatter", grid=True, x="GDP per capita (USD)", y="Life satisfaction")
plt.axis([23_500, 62_500, 4, 9])
plt.show()
```

```python
# Defining the models
model_based_model = LinearRegression() # Represents the Model-Based Learning
instance_based_model = KNeighborsRegressor(n_neighbors=3) # Represents the Instance-Based Learning
```

```python
# Training
model_based_model.fit(x, y)
instance_based_model.fit(x, y)
```

```python
x_new = [[59_687.49]] # Brazil´s GPD per capita in 2025
```

```python
model_based_model.predict(x_new) # Predicting new instances based on the train
```

```python
instance_based_model.predict(x_new) # Predicting new instances based on the known instances
```
